# Langchain 처리방식
invoke(하나씩) / batch(한꺼번에) / stream()

```python
from dotenv import load_dotenv
import os

OPENAI_API_KEY = os.environ['OPENAI_API_KEY']
#PINECONE_API_KEY = os.environ['PINECONE_API_KEY']
print("API 키 로드 완료")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(model="gpt-4o-mini")

# invoke() - 단일 입력, 단일 응답 (동기)
response = llm.invoke("삼성전자 주가에 영향을 주는 요인은?")
print(response.content)
```

```python
# batch() - 여러 입력을 한 번에 처리 (동기)
# 중복 토큰을 줄여 묶어 처리 → 비용 절감 효과
questions = [
    "삼성전자 주가에 영향을 주는 요인은?",
    "코스피 지수란 무엇인가?",
    "ETF와 펀드의 차이는?"
]
responses = llm.batch(questions)
for r in responses:
    print(r.content)
    print()
```

```python
messages = [
    ("system", "당신은 탁월한 주식 애널리스트 입니다. 현재 당신의 수익률은 대한민국 최고 입니다."),
    ("human", "현재 중동에 전쟁으로 유가가 치솟고 있습니다. 이러한 상황에서 피해야할 주식과 매입해야할 주식을 분석해서 top5를 알려주세요. 각 업체명과 함께 최근 매출, 선정한 이유, 리스크요인 등을 함께 알려주세요.") 
]

response = llm.invoke(messages)
print(response.content)
```

```python
# stream() - 실시간 스트리밍 출력 (토큰 단위로 출력)
for chunk in llm.stream("환율이 수출 기업에 미치는 영향은?"):
    print(chunk.content, end="", flush=True)
```

# 동기, 비동기 처리

```python
import asyncio
import time
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(model="gpt-4o-mini")

questions = [f"주식 투자 팁 {i}번째" for i in range(10)]

# 비동기 병렬 처리
async def invoke_async(question):  # invoke 1번
    response = await llm.ainvoke(question)
    return response.content

async def invoke_parallel(questions):   #parallel 여러번
    tasks = [invoke_async(q) for q in questions]
    return await asyncio.gather(*tasks)
```

```python
res=await invoke_async(questions[0])
print(res)
```

```python
# 비동기 실행 시간 측정
start = time.perf_counter()
results = await invoke_parallel(questions)
end = time.perf_counter()
print(f"비동기 처리: {end - start:.2f}초")   # 약 5~6초
```

```python
results
```

```python

for index, content in enumerate(results):
    print(f"{index}, content:{content}")
```

```python
# 동기 실행 시간 측정
start = time.perf_counter()
for q in questions:
    response = llm.invoke(q)
end = time.perf_counter()
print(f"동기 처리: {end - start:.2f}초")     # 약 40~50초
```

```python
from langchain_community.callbacks import get_openai_callback

# 토큰 사용량
with get_openai_callback() as callback:
    message = [HumanMessage(content="서울 광장시장에서 가장 맛있는 길거리 음식은?")]
    response = llm.invoke(message)
    print(response.content)
    print(f"\n총 토큰: {callback.total_tokens}")
    print(f"프롬프트 토큰: {callback.prompt_tokens}")
    print(f"응답 토큰: {callback.completion_tokens}")
    print(f"비용(USD): ${callback.total_cost:.6f}")
```

# 랭체인의 대화 히스토리 관리 객체

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory

llm = ChatOpenAI(model="gpt-4o-mini")

prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 주식 투자 전문가입니다."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])

chain = prompt | llm
```

```python
# 세션별 히스토리 저장소
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

chain_with_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history"
)
```

```python
# 첫 번째 질문 (runnable chain)
response = chain_with_history.invoke(
    {"input": "삼성전자 주식을 지금 매수해도 될까요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```

```python
# 후속 질문 - 같은 session_id로 맥락 유지
response = chain_with_history.invoke(
    {"input": "그렇다면 목표 주가는 얼마로 보시나요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```

```python
# 후속 질문 - 같은 session_id로 맥락 유지
response = chain_with_history.invoke(
    {"input": "그렇다면 주가의 위험요소는 무엇인가요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```
