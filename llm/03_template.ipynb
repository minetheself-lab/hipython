# 프롬프트 구성 > LCEL 파이프라인 구성 > invoke

금융도메인 프롬프트 : system(역할, 페르소나), user, ai(퓨샷러닝), user 로 구성합니다.


```python
### 코드 예시: 백종원 말투 흉내내기 (Few-shot Learning)
from langchain_openai import ChatOpenAI
from langchain_core.prompts  import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# temperature를 약간 높이면(0.8) 표현이 더 자유롭고 창의적이게 됨
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.8)

# - System: 모델의 기본 역할 지정
# - 예시 대화(few-shot examples): 실제 말투 패턴을 보여줌
# - User: 새로운 질문
prompt = ChatPromptTemplate.from_messages([
    ("system", "너는 금융 전문가이자, 백종원 대표의 말투를 자연스럽게 흉내내는 AI야. 충청도 사투리와 친근한 말투로 대답해."),

    # [예시 1] — 사용자가 물어봤고, 백종원 말투로 답변한 사례
    ("user", "ETF 상품을 고르는 비법을 알려주세요."),
    ("ai", "그라믄 말이여~ 좋은 상품을 고르려면, 많이 찾아보는 것이 포인트여. 그래야 손실을 줄여준당게잉."),

    # [예시 2] — 또 다른 백종원식 대답 예시
    ("user", "김치찌개 맛있게 끓이는 법은요?"),
    ("ai", "김치찌개는 묵은지 써야혀~ 그래야 깊은 맛이 나. 돼지고기 좀 넣어주면 금상첨화여잉."),

    # [새로운 요청] — 실행 시 전달될 질문 변수. 이제 모델이 같은 말투로 이어서 대답해야 함
    # ("user", "{question}")]
    # [새로운 요청] - 3개의 추가 파라미터 활용
    ("user", 
          "대상: {target}\n"
          "설명 수준: {level}\n"
          "질문: {question}"
    )
])

# 출력 파서 (응답 텍스트만 추출)
parser = StrOutputParser()

# LCEL 파이프라인 구성 (Prompt → LLM → Parser)
chain = prompt | llm | parser
```

```python
# 실행 — invoke()에 직접 새로운 질문 전달
result = chain.invoke({"question": "좋은 펀드를 추천해주세요"})
print(result)
```

```python
chain_input = {
    "market_type": "코스피(KOSPI)",
    "target": "이제 막 주식 시작한 주린이",
    "level": "비유를 섞어서 아주 쉽게",
    "question": "요즘 삼성전자 왜 이려?"
}
result = chain.invoke(chain_input)

print(result)
```
