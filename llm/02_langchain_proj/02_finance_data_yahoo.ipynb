```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

llm = ChatOpenAI(model="gpt-4o", temperature=0)
```

# llm 보고서 생성하기

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers  import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ("system", """
     Want assistance provided by qualified individuals enabled with experience on understanding charts 
     using technical analysis tools while interpreting macroeconomic environment prevailing across world consequently 
     assisting customers acquire long term advantages requires clear verdicts therefore seeking same through informed 
     predictions written down precisely! First statement contains following content- 
     “Can you tell us what future stock market looks like based upon current conditions ?"."""),
    ("user", """
        {company} 주식에 투자를 해도 될까요? 마크다운 형식의 투자 보고서를 한글로 작성해 주세요.
     """)
])

output_parser = StrOutputParser()

chain = prompt | llm | output_parser

company = "Microsoft"

response= chain.invoke({"company": company})
print(response)
```

# 기본 정보 출력

```python
import yfinance as yf

# Microsoft 주식 심볼 (MSFT)
ticker = 'MSFT'

# 티커 정보 가져오기
msft = yf.Ticker(ticker)

# 기본 정보 조회
basic_info = {
    "회사명": msft.info.get('longName'),
    "산업": msft.info.get('industry'),
    "부문": msft.info.get('sector'),
    "시가총액": msft.info.get('marketCap'),
    "총 발행 주식 수": msft.info.get('sharesOutstanding'),
    "본사 위치": msft.info.get('country'),
    "CEO": msft.info.get('ceo'),  # 일부 주식에서 사용 가능
    "웹사이트": msft.info.get('website'),
}

# 기본 정보 출력
for key, value in basic_info.items():
    print(f"{key}: {value}")
```

# 과거 주가데이터 정보

# 재무제표 정보 가져오기

```python
# 재무제표 정보 가져오기
income_statement = msft.financials  # 손익계산서
balance_sheet = msft.balance_sheet  # 대차대조표
cashflow_statement = msft.cashflow  # 현금흐름표
```

# 주요 지표 조회

```python
# 주요 지표 조회
pe_ratio = msft.info['forwardPE']           # P/E 비율 (예상)
pb_ratio = msft.info['priceToBook']         # P/B 비율
eps = msft.info['trailingEps']              # EPS (주당순이익, 최근 12개월)
roe = msft.info['returnOnEquity']           # ROE (자기자본이익률)

# 데이터 출력
print(f"P/E Ratio (Forward): {pe_ratio}")
print(f"P/B Ratio: {pb_ratio}")
print(f"EPS (Trailing 12 Months): {eps}")
print(f"ROE: {roe}")
```

# 배당수익률 조회

```python
# 배당수익률 조회
dividend_yield = msft.info.get('dividendYield')

# 배당수익률 출력 (퍼센트로 변환)
if dividend_yield:
    dividend_yield_percent = dividend_yield * 100
    print(f"배당수익률: {dividend_yield_percent:.2f}%")
else:
    print("Microsoft는 현재 배당을 지급하지 않습니다.")
```

# 현재 거래량 조회 (최근 거래일 기준)

```python
# 현재 거래량 조회 (최근 거래일 기준)
current_volume = msft.info.get('volume')

# 최근 거래량 출력
print(f"현재 거래량: {current_volume}")
```

```python
# 과거 1개월 간의 주가 및 거래량 데이터 가져오기
historical_data = msft.history(period="1mo")

# 거래량 데이터 출력
print(historical_data['Volume'])
```
