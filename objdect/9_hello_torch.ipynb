# iris
((150, 4)는 2차원, (150,))는 1차원

```python
import torch
import torch.nn as nn
import torch.optim as optim

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from torch.utils.data import TensorDataset, DataLoader
```

```python
# 1. 재현성 고정
torch.manual_seed(1)

# 2. 데이터 로드
iris=load_iris()
X=iris.data
y=iris.target
X.shape, y.shape
```

```python
# 3. train/test 분할
X_train, X_test, y_train, y_test=train_test_split(
  X, y, test_size=0.3, random_state=1, stratify=y
  
)
```

```python
# 4. X값 스케일링- array, 사이킷런 라이브러리
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
#5. Tensor데이터 변환(ndarray->tensor)
X_train=torch.tensor(X_train, dtype=torch.float32)
X_test=torch.tensor(X_test, dtype=torch.float32)
y_train=torch.tensor(y_train, dtype=torch.long)
y_test=torch.tensor(y_test, dtype=torch.long)
```

```python
#6. 데이터셋/ 데이터로더
train_dataset=TensorDataset(X_train, y_train)
test_dataset=TensorDataset(X_test, y_test)

train_loader=DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader=DataLoader(test_dataset, batch_size=16, shuffle=False)
```

# 신경망 모델 모듈, 객체지향
기본 모듈에 추가로 입력.nn 모듈안에 함수들이 만들어져 있다. 

```python
#7. 신경망 모델 정의
class IrisNet(nn.Module):
  def __init__(self): # 신경망 모델 생성자
      super().__init__()
      self.fc1=nn.Linear(4,16) #입력특성 4개
      self.relu=nn.ReLU()
      self.fc2=nn.Linear(16,3) #출력 클래스 3개
      
  def forward(self, x): #순방향학습
      x=self.fc1(x)
      x=self.relu(x)
      x=self.fc2(x)
      return x
```

```python
model=IrisNet()
model
```

```python
#8.손실함수와 옵티마이저
criterion=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters(), lr=0.01)
```

```python
#9. 학습루프
num_epochs=100
loss_history=[]

for epoch in range(num_epochs):
    model.train() #학습
    running_loss=0.0
    
    for x_batch, y_batch in train_loader:
        optimizer.zero_grad()  #기울기 초기화
        outputs=model(x_batch)  #순전파
        loss=criterion(outputs, y_batch) #손실계산
        loss.backward()   #역전파
        optimizer.step()   #가중치 업데이트
        
        running_loss +=loss.item()
        
    epoch_loss= running_loss/len(train_loader)
    loss_history.append(epoch_loss) 
    
    if (epoch+1) %10==0:
       print(f"Epoch [{epoch+1}/{num_epochs}], Loss:{epoch_loss:.4f}")     
```

```python
#10. 평가
model.eval()
correct=0
total=0

with torch.no_grad():
    for x_batch, y_batch in test_loader:
        outputs=model(x_batch)
        _, predicted=torch.max(outputs, 1)
        total +=y_batch.size(0)
        correct +=(predicted==y_batch).sum().item()
        
accuracy=100*correct/total
print(f"Test Accuracy: {accuracy:.2f}%")
```

```python
import matplotlib.pyplot as plt
plt.plot(loss_history)
plt.title('Train Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.show()
```

```python
#9. 학습루프 학습횟수 조절
num_epochs=200
loss_history=[]

for epoch in range(num_epochs):
    model.train() #학습
    running_loss=0.0
    
    for x_batch, y_batch in train_loader:
        optimizer.zero_grad()  #기울기 초기화
        outputs=model(x_batch)  #순전파
        loss=criterion(outputs, y_batch) #손실계산
        loss.backward()   #역전파
        optimizer.step()   #가중치 업데이트
        
        running_loss +=loss.item()
        
    epoch_loss= running_loss/len(train_loader)
    loss_history.append(epoch_loss) 
    
    if (epoch+1) %10==0:
       print(f"Epoch [{epoch+1}/{num_epochs}], Loss:{epoch_loss:.4f}")     
```

```python
#10. 평가
model.eval()
correct=0
total=0

with torch.no_grad():
    for x_batch, y_batch in test_loader:
        outputs=model(x_batch)
        _, predicted=torch.max(outputs, 1)
        total +=y_batch.size(0)
        correct +=(predicted==y_batch).sum().item()
        
accuracy=100*correct/total
print(f"Test Accuracy: {accuracy:.2f}%")
```

# Mnist 신경망

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
# 2. 데이터셋 로드
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.5,), (0.5,))])
train_dataset = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root="./data", train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)
```

```python
# 3. 모델 정의
import torch.nn as nn
import torch.nn.functional as F

class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(28*28, 256)
        self.fc2 = nn.Linear(256, 128)
        self.fc3 = nn.Linear(128, 10)

    def forward(self, x):
        x = x.view(-1, 28*28)  # Flatten
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        return self.fc3(x)

model = MLP().to(device)
```

```python
# 4. 손실함수와 옵티마이저
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
# 5. 학습 루프
epochs = 5
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for X, y in train_loader:
        X, y = X.to(device), y.to(device)
        optimizer.zero_grad()
        outputs = model(X)
        loss = criterion(outputs, y)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    print(f"Epoch {epoch+1}/{epochs}, Loss: {running_loss/len(train_loader):.4f}")
```

```python
# 6. 평가
model.eval()
correct, total = 0, 0
with torch.no_grad():
    for X, y in test_loader:
        X, y = X.to(device), y.to(device)
        outputs = model(X)
        _, preds = torch.max(outputs, 1)
        correct += (preds == y).sum().item()
        total += y.size(0)

print(f"Test Accuracy: {100 * correct / total:.2f}%")
```
